```python
# Importing Libraries
import zipfile
import pandas as pd
import os

# Extracting File and Loading Dataset
zip_path = "netflix-shows.zip"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("C:/Users/hp/Desktop/Yohana Tariku/Data_Cleaning_Project")

csv_path = "C:/Users/hp/Desktop/Yohana Tariku/Data_Cleaning_Project/netflix_titles.csv"

if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
else:
    raise FileNotFoundError(f"{csv_path} does not exist")
```

```python
# Load the dataset
df_cleaned = pd.read_excel("netflix_movies_cleaned_v1_20251222.xlsx")

# Rename columns for clarity and consistency
df_cleaned.rename(columns={
    "type": "content_type",
    "listed_in": "genres"
}, inplace=True)

# Ensure all column names are lowercase and stripped
df_cleaned.columns = df_cleaned.columns.str.lower().str.strip()

# Save the updated dataset
df_cleaned.to_excel("netflix_movies_cleaned_v2_20251222.xlsx", index=False)
```
